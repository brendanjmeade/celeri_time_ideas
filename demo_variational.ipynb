```python
def admm_state_estimation(
    A, b, x_initial, lambda_reg, rho=1.0, tol=1e-4, max_iters=1000
):
    """
    Maybe (?) solve the state estimation problem via ADMM.

    Hessian calculation is a mess/sketchy?

    Not sure how to generalize to coupling.

    Parameters:
      A         : (obs_dim x state_dim) linear operator.
      b         : (T x obs_dim) observed data.
      x_initial : (state_dim,) user-specified initial condition for time t=0.
      lambda_reg : (state_dim,) vector of smoothing weights.
      rho       : ADMM penalty parameter (default: 1.0).
      tol       : Tolerance for convergence (default: 1e-4).
      max_iters : Maximum number of ADMM iterations (default: 1000).

    Returns:
      x_est : (T x state_dim) estimated state trajectory.
    """
    T = b.shape[0]  # Number of time steps
    state_dim = A.shape[1]  # Number of state variables
    n = T * state_dim  # Total decision variables

    # Create D = diag(lambda_reg)
    D = np.diag(lambda_reg)

    # Build Hessian
    Q = np.zeros((n, n))
    data_block = 2 * (A.T @ A)
    for t in range(T):
        row_start = t * state_dim
        row_end = row_start + state_dim
        Q[row_start:row_end, row_start:row_end] = data_block.copy()
        if t == 0:
            Q[row_start:row_end, row_start:row_end] += 2 * D
        elif t == T - 1:
            Q[row_start:row_end, row_start:row_end] += 2 * D
        else:
            Q[row_start:row_end, row_start:row_end] += 4 * D
    # Off-diagonal contributions from smoothness:
    for t in range(1, T):
        idx1 = (t - 1) * state_dim
        idx2 = t * state_dim
        Q[idx1 : idx1 + state_dim, idx2 : idx2 + state_dim] = -2 * D
        Q[idx2 : idx2 + state_dim, idx1 : idx1 + state_dim] = -2 * D

    # Build the linear term vector d.
    d = np.zeros(n)
    for t in range(T):
        d[t * state_dim : (t + 1) * state_dim] = 2 * (A.T @ b[t])

    # Precompute the matrix for the x-update, M = Q + rho I.
    M = Q + rho * np.eye(n)
    M_factor = cho_factor(M)

    # Initial guess
    x0_matrix = np.zeros((T, state_dim))
    A_pinv = np.linalg.pinv(A)
    for t in range(T):
        x0_matrix[t] = A_pinv.dot(b[t])
        x0_matrix[t] = np.maximum(x0_matrix[t], 0) + 1e-3
    x0_matrix[0] = x_initial
    x = x0_matrix.flatten()

    # Initialize ADMM auxiliary variables
    z = x.copy()
    u = np.zeros(n)

    # Define projection onto the constraints:
    # 1. Each x(t) must be non-negative.
    # 2. The initial condition must be satisfied: z(0) = x_initial
    def project(z_vec):
        Z = z_vec.reshape((T, state_dim))

        # Force C to be zero or positive (non-negative constraint)
        Z = np.maximum(Z, 0)
        Z[0] = x_initial
        return Z.flatten()

    # ADMM iterations.
    for k in range(max_iters):
        # x-update: solve (Q + rho I)x = d + rho*(z - u)
        v = z - u
        rhs = d + rho * v
        x_new = cho_solve(M_factor, rhs)

        # z-update: projection onto the feasible set.
        z_new = project(x_new + u)

        # Dual variable update.
        u = u + x_new - z_new

        # Compute residuals.
        r_norm = np.linalg.norm(x_new - z_new)
        s_norm = np.linalg.norm(rho * (z_new - z))

        if r_norm < tol and s_norm < tol:
            print(f"ADMM converged in {k} iterations.")
            break

        x, z = x_new, z_new
    else:
        print("ADMM reached the maximum number of iterations.")

    x_est = x_new.reshape((T, state_dim))
    return x_est
```

```python
# Define observation coordinates
n_pts = 50
x_lim = 50
x = np.linspace(-x_lim, x_lim, n_pts)


def get_velocity(slip_rate, x, locking_depth):
    velocity = slip_rate / np.pi * np.arctan(x / locking_depth)
    return velocity


# Define fault patches
n_patches = 2
min_depth = 1e-10
max_depth = 30
locking_depths = np.linspace(min_depth, max_depth, n_patches + 1)

# Velocities
s = 1.0
v_inter = s / np.pi * np.arctan(x / max_depth)
v_mat = np.zeros((n_patches, n_pts))

# Calculate elastic deformation from each patch
for i in range(len(locking_depths) - 1):
    # Velocity kernel calcuation
    v_upper = get_velocity(1, x, locking_depths[i])
    v_lower = get_velocity(1, x, locking_depths[i + 1])
    v = v_lower - v_upper
    v_mat[i, :] = v

# Generate synthetic time series
nt = 100
t_coseismic = 0.5
# noise_scale = 0.000
noise_scale = 0.002


t = np.linspace(0, 1, nt)
t_vec = np.copy(t)
dt = t[1] - t[0]
v0 = 0.5 * np.ones(nt)
v1 = np.zeros(nt)
v2 = np.zeros(nt)

# Coseismic jump on upper fault patch
v_coseismic = 25
coseismic_idx = np.argmin(np.abs(t - t_coseismic))
v1[coseismic_idx] = v_coseismic

# Shallow slow slip event before earthquake
sse_amplitude = 3.0
sse_center = 0.35
sse_width = 0.02
v_sse = sse_amplitude * np.exp(-((t - sse_center) ** 2) / (2 * sse_width**2))
v1 += v_sse

# Deep slow slip event before earthquake
sse_amplitude = 1.0
sse_center = 0.25
sse_width = 0.03
v_sse = sse_amplitude * np.exp(-((t - sse_center) ** 2) / (2 * sse_width**2))
v2 += v_sse

# Change interseismic with earthquake
v0[coseismic_idx + 1 : :] = 0.25

# Afterslip on lower fault patch
afterslip_amplitude = 200
# afterslip_amplitude = 0

afterslip_time_scale = 8.0
v2[t > t_coseismic] = afterslip_amplitude * np.exp(
    -afterslip_time_scale * t[t > t_coseismic] - t_coseismic
)

# Assemble linear operator (parameters -> data)
A = np.zeros((n_pts, 3))
A[:, 0] = v_inter
A[:, 1] = -v_mat[0, :]
A[:, 2] = -v_mat[1, :]

x_mat = np.zeros((n_pts, nt))
for i in range(nt):
    # x_mat[:, i] = 0 * v_inter - (v1[i] * v_mat[0, :]) - (v2[i] * v_mat[1, :])
    x_mat[:, i] = v0[i] * v_inter - (v1[i] * v_mat[0, :]) - (v2[i] * v_mat[1, :])

b_true = np.cumsum(x_mat, axis=1) * dt

# Add synthetic noise
b_noise = b_true + noise_scale * np.random.normal(size=b_true.shape)

# Packaging
x0 = np.cumsum(v0)
x1 = np.cumsum(v1)
x2 = np.cumsum(v2)

b_true_noco = np.copy(b_true)
b_noise_noco = np.copy(b_noise)
x1_noco = np.copy(x1)
coseismic_step = x1_noco[coseismic_idx] - x1_noco[coseismic_idx - 1]
x1_noco[coseismic_idx::] -= coseismic_step
x0_noco = np.copy(x0)
x2_noco = np.copy(x2)

for i in range(n_pts):
    # Remove coseismic from noise free observations
    coseismic_step = b_true_noco[i, coseismic_idx] - b_true_noco[i, coseismic_idx - 1]
    b_true_noco[i, coseismic_idx::] -= coseismic_step

    # Remove coseismic from noisy observations
    coseismic_step = b_noise_noco[i, coseismic_idx] - b_noise_noco[i, coseismic_idx - 1]
    b_noise_noco[i, coseismic_idx::] -= coseismic_step

b = np.copy(b_noise_noco.T)

# Block linear operator
v_block = np.zeros_like(x)
v_block[x < 0] = -0.5
v_block[x > 0] = 0.5

AA = np.zeros_like(A)
# AA[:, 0] = v_block
# AA[:, 1] = -v_mat[0, :]
# AA[:, 2] = -v_mat[1, :]

AA[:, 0] = v_block + v_mat[0, :] + v_mat[1, :]
AA[:, 1] = -v_mat[0, :]
AA[:, 2] = -v_mat[1, :]


# Linear operator for block case
A_block = np.zeros_like(A)
A_block[:, 0] = v_block
A_block[:, 1] = v_mat[0, :]
A_block[:, 2] = v_mat[1, :]
```

```python
def velocity_to_coupling(ve, vk):
    # Avoid division by zero by leaving t==0 untouched.
    coupling = np.zeros_like(ve)
    nonzero = vk != 0
    coupling[nonzero] = ve[nonzero] / vk[nonzero]
    return coupling


def coupling_to_velocity(coupling, vk):
    return coupling * vk


def admm_state_estimation_ramp(
    A, b, x_initial, lambda_reg, rho=1.0, tol=1e-4, max_iters=1000
):
    """
    Solve the state estimation problem with a constant ramp parameter using ADMM.

    The state at time t is parameterized as:
        x(t) = [ alpha * t,  x1(t),  x2(t) ]
    where alpha is a single global parameter descibing the block rate

    The objective minimizes:
       sum_{t=0}^{T-1} || A * [alpha*t, x1(t), x2(t)] - b[t] ||^2 +
       lambda_reg * sum_{t=1}^{T-1} [ (x1(t)-x1(t-1))^2 + (x2(t)-x2(t-1))^2 ]

    subject to a coupling constraint defined by the gradient:
         coupling for x1(t) = (x1(t)-x1(t-1))/(alpha)
         coupling for x2(t) = (x2(t)-x2(t-1))/(alpha)
    which is forced to lie within specified bounds for t>=1

    Parameters:
      A         : (obs_dim x 3) known matrix
      b         : (T x obs_dim) observed data
      x_initial : (3,) initial state at t=0.  Here x_initial[0] is the initial ramp value
      lambda_reg : scalar smoothing weight (applied to x1 and x2). Should generlize as separate for each
      rho       : ADMM penalty parameter
      tol       : Convergence tolerance
      max_iters : Maximum ADMM iterations

    Returns:
      x_est : (T x 3) estimated state trajectory, where
             x_est[t] = [ alpha * t,  x1(t),  x2(t) ]
    """
    T = b.shape[0]
    # Decision variable theta has size: 1 (for alpha) + 2*T (for x1 and x2)
    # NOTE: I hate this shaping and need to generalize this.
    n = 1 + 2 * T

    # Precompute A^T@A (3x3 matrix)
    B = A.T @ A

    # Build the Hessian, Q, and linear term d
    Q = np.zeros((n, n))
    d = np.zeros(n)

    # For each time step, the model state is [alpha*t, x1, x2]
    # We treat the data term at time t as
    # || M(t)*[alpha, x1, x2]^T - b[t] ||^2,
    # with M(t) having columns [t*A[:,0], A[:,1], A[:,2]]
    # NOTE:  This is the magic.  Seperating the linear term from the others was brutal
    for t in range(T):
        # Indices in theta: 0 for alpha; 1..T for x1; (T+1)..(1+2T) for x2
        i_alpha = 0
        i_x1 = 1 + t
        i_x2 = 1 + T + t

        # Hessian contributions from data fit (multiplied by 2)
        Q[i_alpha, i_alpha] += 2 * (t**2) * B[0, 0]
        Q[i_alpha, i_x1] += 2 * t * B[0, 1]
        Q[i_alpha, i_x2] += 2 * t * B[0, 2]
        Q[i_x1, i_alpha] += 2 * t * B[1, 0]
        Q[i_x1, i_x1] += 2 * B[1, 1]
        Q[i_x1, i_x2] += 2 * B[1, 2]
        Q[i_x2, i_alpha] += 2 * t * B[2, 0]
        Q[i_x2, i_x1] += 2 * B[2, 1]
        Q[i_x2, i_x2] += 2 * B[2, 2]

        # Linear term from data fit: -2 * (M(t)^T @ b[t])
        Abt = A.T @ b[t]
        d[i_alpha] += -2 * t * Abt[0]
        d[i_x1] += -2 * Abt[1]
        d[i_x2] += -2 * Abt[2]

    # Add smoothness terms (only for x1 and x2) for t = 1,..., T-1
    for t in range(1, T):
        # For x1:
        i_prev = 1 + (t - 1)
        i_curr = 1 + t
        Q[i_prev, i_prev] += lambda_reg
        Q[i_curr, i_curr] += lambda_reg
        Q[i_prev, i_curr] += -lambda_reg
        Q[i_curr, i_prev] += -lambda_reg

        # For x2:
        i_prev = 1 + T + (t - 1)
        i_curr = 1 + T + t
        Q[i_prev, i_prev] += lambda_reg
        Q[i_curr, i_curr] += lambda_reg
        Q[i_prev, i_curr] += -lambda_reg
        Q[i_curr, i_prev] += -lambda_reg

    # Precompute the matrix for the x-update: M_mat = Q + rho I.
    M_mat = Q + rho * np.eye(n)
    M_factor = cho_factor(M_mat)

    # Build an initial guess for theta = [alpha, x1(0)...x1(T-1), x2(0)...x2(T-1)]
    theta0 = np.zeros(n)
    theta0[0] = x_initial[0]  # initial alpha
    theta0[1] = x_initial[1]  # x1 at t=0
    theta0[1 + T] = x_initial[2]  # x2 at t=0

    # TODO: These should all be replaced with something cheaper
    A_pinv = np.linalg.pinv(A)
    for t in range(T):
        est = A_pinv @ b[t]
        if t > 0:
            theta0[1 + t] = est[1]
            theta0[1 + T + t] = est[2]

    # Initialize ADMM variables
    x = theta0.copy()
    z = theta0.copy()
    u = np.zeros(n)

    # Projection function
    # For t=0, the state is fixed to x_initial
    # For t>=1, we enforce the coupling (gradient) constraints
    # (x1(t)-x1(t-1))/(alpha) and (x2(t)-x2(t-1))/(alpha)
    # are clipped to lie between min_coupling and max_coupling
    # TODO: Need to pass limits rather than hardcode
    # NOTE: Is there a smoother way to do this so that things get closer without jumping
    # NOTE: Need to generalize indexing for arbitrary number of variables.

    def project(theta_vec):
        theta_proj = theta_vec.copy()
        alpha = theta_proj[0]

        # Apply initial condition
        theta_proj[1] = x_initial[1]  # x[0]
        theta_proj[1 + T] = x_initial[2]  # x2[0]
        min_coupling = -20.0
        max_coupling = 1.0
        # For t >= 1, project the differences.
        # For x1:
        for t in range(1, T):
            idx_prev = 1 + (t - 1)
            idx_curr = 1 + t
            diff = theta_proj[idx_curr] - theta_proj[idx_prev]
            # Denom is alpha; avoid division by (nearly) zero.
            if np.abs(alpha) < 1e-8:
                continue
            coupling = diff / alpha
            coupling_proj = np.clip(coupling, min_coupling, max_coupling)
            theta_proj[idx_curr] = theta_proj[idx_prev] + coupling_proj * alpha
        # For x2:
        for t in range(1, T):
            idx_prev = 1 + T + (t - 1)
            idx_curr = 1 + T + t
            diff = theta_proj[idx_curr] - theta_proj[idx_prev]
            if np.abs(alpha) < 1e-8:
                continue
            coupling = diff / alpha
            coupling_proj = np.clip(coupling, min_coupling, max_coupling)
            theta_proj[idx_curr] = theta_proj[idx_prev] + coupling_proj * alpha

        return theta_proj

    # ADMM iterative solve
    for k in range(max_iters):
        # x-update: solve (Q + rho I) x = d + rho*(z - u)
        rhs = d + rho * (z - u)
        x_new = cho_solve(M_factor, rhs)

        # z-update: project onto the feasible set.
        z_new = project(x_new + u)

        # Dual variable update.
        u = u + x_new - z_new

        r_norm = np.linalg.norm(x_new - z_new)
        s_norm = np.linalg.norm(rho * (z_new - z))
        if r_norm < tol and s_norm < tol:
            print(f"ADMM converged in {k} iterations.")
            x_new = z_new  # ensure consistency
            break

        x, z = x_new, z_new
    else:
        print("ADMM reached maximum iterations.")

    # Extract estimated parameters.
    alpha_est = x_new[0]
    x1_est = x_new[1 : 1 + T]
    x2_est = x_new[1 + T : 1 + 2 * T]

    # Build the state trajectory.
    x_est = np.zeros((T, 3))
    for t in range(T):
        x_est[t, 0] = alpha_est * t
        x_est[t, 1] = x1_est[t]
        x_est[t, 2] = x2_est[t]

    return x_est
```
